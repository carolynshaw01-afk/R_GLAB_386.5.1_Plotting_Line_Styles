##**Example 2: Dotted Line style**

### We can use **':'** or **dotted** value in Linestyle argument.

```python
# Carolyn's Self-Talk:
# import = "Bring a tool/library into my notebook so I can use it."
# matplotlib = the larger Python data-visualization library.  Think:  the whole toolbox for making graphs.
# . = "Go inside." | Dot = Go inside and get something.
# pyplot = the part inside Matplotlib that contains many of the graph-making commanfs I am using.
# So, matplotlib.pyplot means:  "Go inside Matplotlib and get pyplot."
# as = "I'm going to give this a shorter nickname."
# plt = the nickname we give matplotlib.pyplot.
# So the WHOLE LINE says:  🧠 "Bring in Matplotlib's graphing tools, and I'll call them plt for short."
import matplotlib.pyplot as plt
import numpy as np
# Carolyn's Self-Talk:
# import = bring in a library so I can use it.
# numpy = a Python library used for working with numbers, arrays, and numberical data.
# as = give it a shorter nickname.
# np = the nickname for NumPy.  🧠 "Bring in NumPy, and I'll call it np for short."

xpoints = np.array([1, 8])
ypoints = np.array([3, 10])

# 🧠 Carolyn's Self-Talk:
# plt → “Go to Matplotlib's plotting tools.”
# .  → “Go inside and grab a tool.”
# plot  → “I want you to make a plot.”
# (  → “Here are my instructions.”
# xpoints  → “Use these values across my X-axis: [1, 8].”  ====>  The [1, 8] and [3, 10] do not cause the dots.  ====> Those numbers determine where the line goes.
#  The ":" determines how the line looks.
# ,  → “And...”
# ypoints
# → “Use these values on my Y-axis: [3, 10].”
# So, Python pairs them:  (1, 3) and (8, 10)  then:  
# linestyle = ":"    ====>  That's the instruction responsible for what your eyes see on the graph: dot dot dot dot dot.
#  → “Draw the line using dots.”  Why does the graph have dots?  Because I specifically told Matplotlib how I wanted the line to look with linestyle = ":".
# And finally:
# plt.show()
# then:
# # → “Show me the finished graph.”

plt.plot(xpoints, ypoints, linestyle = ":")
plt.show()
```

##**Example 3: Dash Line style**

### We can use **'--'** or  **dashed** value in Linestyle argument.

```python
# Carolyn's Self-Talk:
# import = "Bring a tool/library into my notebook so I can use it."
# matplotlib = the larger Python data-visualization library.  Think:  the whole toolbox for making graphs.
# . = "Go inside." | Dot = Go inside and get something.
# pyplot = the part inside Matplotlib that contains many of the graph-making commanfs I am using.
# So, matplotlib.pyplot means:  "Go inside Matplotlib and get pyplot."
# as = "I'm going to give this a shorter nickname."
# plt = the nickname we give matplotlib.pyplot.
# So the WHOLE LINE says:  🧠 "Bring in Matplotlib's graphing tools, and I'll call them plt for short."
import matplotlib.pyplot as plt
import numpy as np
# import = bring in a library so I can use it.
# numpy = a Python librar used for working with numbers, arrays, and numberical data.
# as = give it a shorter nickname.
# np = the nickname for NumPy.  🧠 "Bring in NumPy, and I'll call it np for short."
# xpoints = the variable name I am creating.  It is the name of the container that will hold my X values.  🧠 "I'm calling this container xpoints because it holds my X points."
# = means =
# Assign/store/put into.  🧠 Read from right to left.  "Take what's on the RIGHT and store it in the name on the LEFT."
# xpoints = :  "Store something in xpoints."
# np = The nickname we gave NumPy when we wrote: import numpy as np, “np means NumPy.”
# . = Go inside 🧠 “Go inside NumPy and use one of its tools.”
# array = The NumPy tool/function we're asking it to use. It creates a NumPy array, which holds a collection of values.  🧠 “NumPy, make me an array.”
# ( = Opens the function call. It means:  🧠 “Here comes the information I'm giving to array.”  We eventually close it with:  )
# Think of the parentheses as the doorway into the function:  array(  put the information in here  )
# [ = array(  put the information in here  ), And the matching:  ]  So, [1, 8] means:  🧠 “Here is my collection of values.”
# 1, 8 = These are the actual X values provided by this R-GLAB example.
# 1  = first X value
# ,  = next value
# 8  = second X value
# Read it slowly:  🧠 “Create a variable called xpoints. Go into NumPy, use its array tool, give it the values 1 and 8, and store that array in xpoints.”
#  Or your super-short version:  xpoints → store → NumPy → array → [1, 8]

xpoints = np.array([1, 8])
ypoints = np.array([3, 10])
# xpoints  → “Use these values across my X-axis: [1, 8].”  ====>  The [1, 8] and [3, 10] do not cause the dots.  ====> Those numbers determine where the line goes.
#  The ":" determines how the line looks.
# ,  → “And...”
# ypoints
# → “Use these values on my Y-axis: [3, 10].”
# So, Python pairs them:  (1, 3) and (8, 10)  then:  
# linestyle = ":"    ====>  That's the instruction responsible for what your eyes see on the graph: dash   dash   dash   dash   dash   dash.
#  → “Draw the dashes.”  Why does the graph have dashes?  Because I specifically told Matplotlib how I wanted the line to look with linestyle = "--".
plt.plot(xpoints, ypoints, linestyle = "--")
# And finally:
# plt.show()
# then:
# # → “Show me the finished graph.”
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# title of 'World Population' blue
# labels growth by year, population in billions green red

years = [1970,1980,1990,1995,2000,2005,2010,2015,2020,2025,2030,2035]
pops = [3.6,4.0,4.4,4.8,5.3,5.7,6.1,6.5,7.3,7.3,7.4,7.5]


# 🧠 Carolyn's Self-Talk  &  my own little Matplotlib decoder. 💡
# linestyle='-.',   # HOW the line is drawn → dash-dot
# color='purple',   # WHAT COLOR the line is → purple
# marker='D'        # means 🧠 “Put a diamond marker at each data point.”  Not simply between pieces of the dash-dot line. Remember your paired data:
#                   (1970, 3.6), (1980, 4.0), (1990, 4.4) ... Each pair gives Matplotlib a data point, and the 'D' says:  “Put a diamond on each one of those points.” 💎
#                   Then the dash-dot line connects those data points.
# markersize=8      # Read the word literally: marker + size.       # markersize=8 = Make each diamond marker 8 points in size. 
# markerfacecoler   # markerfacecolor = What color is INSIDE my marker?  Read it literally:  marker + face + color  |  The face means the inside/fill of your diamond marker. 
#                   # So 'yellow' tells Matplotlib:  🧠 “Fill the inside of each diamond marker with yellow.”
plt.plot(years, pops,
         linestyle='-.',
         color='purple',
         marker='D',
         markersize=8,
         markerfacecolor='yellow')

plt.title('World Population', color='blue')
plt.xlabel('Growth by Year', color='green')
plt.ylabel('Population (billions)', color='red')

plt.show()

# 🧠 Carolyn's Self-Talk:
# Question #1: Which one is your X data — years or pops? A
# Answer   #1: x-axis is years = [1970,1980,1990,1995,2000,2005,2010,2015,2020,2025,2030,2035]
# then
# axis is pops = [3.6,4.0,4.4,4.8,5.3,5.7,6.1,6.5,7.3,7.3,7.4,7.5]
# # And
# Question #2: How can you tell by looking at the graph?
# Answer #2:  X-axis = horizontal = across ⬅️ ➡️
#             y-axis - Vertical   = up and down ⬆️ ⬇️
```
